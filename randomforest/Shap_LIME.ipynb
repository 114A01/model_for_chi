```python
import joblib

# 你的模型檔路徑
package_path = r"C:\Users\pccu1\Desktop\project-model\randomforest\models\complete_model_20260512_131827.joblib"

# 載入模型
loaded_package = joblib.load(package_path)

print("模型載入成功")
```

```python
rf_model = loaded_package['model']

X_val = loaded_package['validation_data']['X']

y_val = loaded_package['validation_data']['y']

feature_names = loaded_package['feature_names']

label_encoder = loaded_package['label_encoder']

print("RF model 載入成功")
print("Validation set shape:", X_val.shape)
```

```python
import shap
import numpy as np

sample = X_val.iloc[[0]]

rf_explainer = shap.TreeExplainer(rf_model)
rf_shap_values = rf_explainer.shap_values(sample)
```

```python
print(type(rf_shap_values))
print(np.array(rf_shap_values).shape)
```

```python
# 模型預測類別
rf_pred = rf_model.predict(sample)[0]

print("預測類別 index:", rf_pred)

# 正確取得 SHAP values
rf_values = rf_shap_values[0][:, rf_pred]

print("SHAP values shape:", rf_values.shape)
```

```python
# 計算重要程度（取絕對值）
rf_importance = np.abs(rf_values)

# 找前五大 feature index
rf_top5_idx = np.argsort(rf_importance)[-5:]

# 找 feature 名稱
rf_top5_features = sample.columns[rf_top5_idx]

print("\nRF Top 5 Features:")

for idx in reversed(rf_top5_idx):
    print(f"{sample.columns[idx]} : {rf_values[idx]:.4f}")
```
